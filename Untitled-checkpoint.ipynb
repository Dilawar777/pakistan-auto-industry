```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import matplotlib.gridspec as gridspec
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_excel("pama.xlsx")
vehicle  = "Vehicles"
typ      = "Column1"
year     = "Year"
units    = "Value"
category = "Category.1"

print(df.head())
```

```python
print("\n" + "=" * 55)
print("COLUMN NAMES")
print("=" * 55)
for col in df.columns:
    print(f"  {col}")
```

```python
print("\n" + "=" * 55)
print("FIRST 5 ROWS")
print("=" * 55)
print(df.head())
```

```python
print("\n" + "=" * 55)
print("DATA TYPES")
print("=" * 55)
print(df.dtypes)
```

```python
print("\n" + "=" * 55)
print("MISSING VALUES")
print("=" * 55)
missing = df.isnull().sum()
if missing.sum() == 0:
    print("No missing values found ✓")
else:
    print(missing[missing > 0])
```

```python
print("\n" + "=" * 55)
print("DUPLICATE ROWS")
print("=" * 55)
dupes = df.duplicated().sum()
print(f"Duplicates found {dupes:,}")
if(dupes > 0):
    df = df.drop_duplicates()
    print(f"Removed. Rows remaining: {df.shape[0]:,}")
```

```python
print("\n" + "=" * 55)
print("UNIQUE VALUES — KEY COLUMNS")
print("=" * 55)
```

```python
print("\n" + "=" * 55)
print("UNIQUE VALUES — KEY COLUMNS")
print("=" * 55)
for col in df.columns:
    unique_count = df[col].nunique()
    if unique_count <= 20:
        print(f"\n{col} ({unique_count} unique):")
        print(f"  {df[col].unique().tolist()}")
    else:
        print(f"\n{col}: {unique_count:,} unique values (too many to list)")
```

```python
df.columns = df.columns.str.strip().str.replace(" " , "_")
print("\n" + "=" * 55)
print("CLEANED COLUMN NAMES")
print("=" * 55)
print(df.columns.tolist())
```

```python
value_col = [c for c in df.columns if "value" in c.lower() or "value" in c]
if value_col:
    df.rename(columns = {value_col[0]: "Units"} , inplace = True)
    print(f"Renamed '{value_col[0]}' → 'Units'")
```

```python
print("\n" + "=" * 55)
print("CREATING DATE COLUMN")
print("=" * 55)
month_col = [c for c in df.columns if "mon" in c.lower() or "Mon" in c]
year_col  = [c for c in df.columns if "ye" in c.lower() or "Ye" in c]
 
if month_col and year_col:
    mc = month_col[0]; yc = year_col[0]
    df["Date"] = pd.to_datetime(
        df[mc].astype(str) + " " + df[yc].astype(str),
        format="%b %Y", errors="coerce"
    )
    nulldates = df["Date"].isnull().sum()
    print(f"Date column created ✓")
    print(f"Date range: {df['Date'].min()} → {df['Date'].max()}")
    if nulldates > 0:
        print(f"Warning: {nulldates} rows couldn't be parsed")
    else:
        print("All dates parsed successfully ✓")
 
```

```python
print("\n" + "=" * 55)
print("UNITS COLUMN STATS")
print("=" * 55)
if "Units" in df.columns:
    print(f"Min  : {df['Units'].min():,}")
    print(f"Max  : {df['Units'].max():,}")
    print(f"Mean : {df['Units'].mean():,.0f}")
    neg = (df["Units"] < 0).sum()
    print(f"Negative values: {neg}")
    if neg > 0:
        df = df[df["Units"] >= 0]
        print(f"Removed {neg} negative rows")
```

```python
df.to_csv("pama_cleaned.csv", index=False)
print("\n" + "=" * 55)
print("STEP 1 COMPLETE")
print("=" * 55)
print(f"  Original rows  : 17,723")
print(f"  Cleaned rows   : {df.shape[0]:,}")
print(f"  Columns        : {df.shape[1]}")
print(f"  Saved to       : pama_cleaned.csv ✓")
```
